```python
df_enroll.groupby("Institution").agg({"Headcount": ["size", "sum"]})
```

```python
df_enroll.groupby(["Institution", "Site"]).agg({"Headcount": ["size", "sum"]})
```

```python
# Use strings "size" and "sum" instead of np.size and np.sum
df_enroll.groupby(["Institution", "Site"]).agg({"Headcount": ["size", "sum"]}).sum()
```

```python
df_enroll[
    (df_enroll["Institution"] == "University Of Utah")
    & (df_enroll["Site"] == "Main Campus")
].groupby(["Institution", "Site", "ZIPCODE"]).agg({"Headcount": ["size", "sum"]})
```

```python
dfTotEnrolByZip = df_enroll.groupby(
    ["Institution", "Site", "ZIPCODE"], as_index=False
).agg(TotEnrol=("Headcount", "sum"))
dfTotEnrolByZip.to_csv(
    os.path.join(int_folder, "campus_totenrol_byzip.csv"), index=False
)
dfTotEnrolByZip
```

```python
dfTotEnrolByZip_College = pd.DataFrame.merge(
    dfCampusCodes, dfTotEnrolByZip, on=["Institution", "Site"]
)
dfTotEnrolByZip_College = dfTotEnrolByZip_College.drop(
    columns=(["Institution", "Site"])
)
dfTotEnrolByZip_College
```

```python
dfTotEnrolByZip_College.to_csv(
    os.path.join(int_folder, "TotEnrolByZipByCollege.csv"), index=False
)
```
