```python
import pandas as pd
import os
import numpy as np
```

```python
working_directory = os.getcwd()
print(working_directory)
```

```python
# Read 2020 csv as pandas dataframe object
data_folder = os.path.join(os.getcwd(), r'data\USHE 2020 Enrollment')
int_folder = os.path.join(os.getcwd(), r'intermediate')
csv_enrol = os.path.join(data_folder, 'headcount.csv')
df_enrol = pd.read_csv(csv_enrol)
```

```python
dfCampusCodes = pd.DataFrame([
                  ['SLCC','Main Campus'                ,'SLCC_MAIN'],
                  ['SLCC','SLCC Larry H. Miller Campus','SLCC_ML'  ],
                  ['SLCC','SLCC South City Center'     ,'SLCC_SC'  ],
                  ['SLCC','SLCC Jordan Campus'         ,'SLCC_JD'  ],
                  ['UU'  ,'Main Campus'                ,'UOFU_MAIN'],
                  ['UVU' ,'Main Campus'                ,'UVU_MAIN' ],
                  ['WSU' ,'Main Campus'                ,'WSU_MAIN' ],
                  ['WSU' ,'WSU Davis Center'           ,'WSU_DAVIS']
                ],columns=['Institution','Site','COLLEGE'])

dfCampusCodes
```

```python
# Show the first 5 rows of a table
df_enrol.head()
```

```python
df_enrol.groupby('Institution').agg({'Headcount': [np.size, np.sum]})
```

```python
df_enrol.groupby(['Institution','Site']).agg({'Headcount': [np.size, np.sum]})
```

```python
df_enrol.groupby(['Institution','Site']).agg({'Headcount': [np.size, np.sum]}).sum()
```

```python
df_enrol[(df_enrol['Institution'] == 'UU') & (df_enrol['Site'] == 'Main Campus')].groupby(['Institution','Site','ZIP_Code']).agg({'Headcount': [np.size, np.sum]})
```

```python
dfTotEnrolByZip = df_enrol.groupby(['Institution','Site','ZIP_Code'],as_index=False).agg(TotEnrol=('Headcount','sum'))
dfTotEnrolByZip.to_csv(os.path.join(int_folder,'campus_totenrol_byzip.csv'),index=False)
dfTotEnrolByZip
```

```python
dfTotEnrolByZip_College = pd.DataFrame.merge(dfCampusCodes,dfTotEnrolByZip,on=['Institution','Site'])
dfTotEnrolByZip_College = dfTotEnrolByZip_College.drop(columns=(['Institution','Site']))
dfTotEnrolByZip_College
```

```python
dfTotEnrolByZip_College.to_csv(os.path.join(int_folder,'TotEnrolByZipByCollege.csv'),index=False)
```
