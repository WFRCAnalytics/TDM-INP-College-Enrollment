```python
#read totals without the join to check against
dfTruckDataSumForChecks = pd.DataFrame()

for dit in dImportTables:

    sqlsumforchecks = """ 
    SELECT
        SUM(od_table.o_d_traffic_sample_trip_counts) AS od_trip_counts_sum,
        SUM(od_table.o_d_traffic_calibrated_trip_volume) AS od_trip_volume_sum,
        COUNT(od_table.mode_of_travel) AS num_records
    FROM
        """ + dit[0] + """ AS od_table
    """

    #display(sqlsumforchecks)

    dfQuerySum = client.query(sqlsumforchecks).to_dataframe()
    dfQuerySum['table'] = dit[0]
    dfTruckDataSumForChecks = dfTruckDataSumForChecks.append(dfQuerySum)

dfTruckDataSumForChecks
```

```python
#create query of all subareaid to subareaid truck flows using join to taz table with subareaid field
dfTruckDataOD = pd.DataFrame()

for dit in dImportTables:
    sql = """
    SELECT
        origin_zone_name,
        destination_zone_name,
        mode_of_travel,
        vehicle_weight,
        day_type,
        day_part,
        data_period,
        o_d_traffic_sample_trip_counts,
        o_d_traffic_calibrated_trip_volume
    FROM
        """ + dit[0] + """ as od_table

    """
    #display(sql)

    dfQuery = client.query(sql).to_dataframe()
    dfTruckDataOD = dfTruckDataOD.append(dfQuery)

dfTruckDataOD
```

```python
# filter by weekday all day
dfTruckDataODfiltered = dfTruckDataOD.loc[(dfTruckDataOD['day_type']=='1: Weekdays (Tu-Th)') & (dfTruckDataOD['day_part']!='0: All Day (12am-12am)')]
dfTruckDataODfiltered
```

```python
display('OD Total Counts: ' + f"{dfTruckDataODfiltered['o_d_traffic_sample_trip_counts'].sum().astype(int):,}")
display('OD Total Volume: ' + f"{dfTruckDataODfiltered['o_d_traffic_calibrated_trip_volume'].sum().astype(int):,}")
```

```python
dfTruckDataODfiltered.groupby(['vehicle_weight']).agg(COUNT=('o_d_traffic_sample_trip_counts','sum'),VOLUME=('o_d_traffic_calibrated_trip_volume','sum'))
```

```python
#dfTruckDataODint = dfTruckDataOD.copy
#dfTruckDataODint['o_d_traffic_calibrated_trip_volume'] = dfTruckDataODint['o_d_traffic_calibrated_trip_volume'].astype(int)
dfTruckDataODfiltered.groupby(['day_type','day_part','data_period','vehicle_weight']).agg(COUNT=('o_d_traffic_sample_trip_counts','sum'),VOLUME=('o_d_traffic_calibrated_trip_volume','sum'))
```
